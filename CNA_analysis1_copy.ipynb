# Linear Regression
What factors play into the annual mean wage of CNAS? 
y will be the annual mean wage for the cnas, aka a_mean


Simple regression: y = a_mean x = total_emp

```python
import statsmodels.api as sm
x = df_cna_by_state_job_stats['jobs_1000']
x = sm.add_constant(x)
y = df_cna_by_state_job_stats['a_mean']
simple_regression_model = sm.OLS(y,x).fit()
print(simple_regression_model.summary())
```

```python
mutliple_x = df_cna_by_state_job_stats[['jobs_1000', 'tot_emp','loc_quotient']]
mutliple_x= sm.add_constant(mutliple_x)
y = df_cna_by_state_job_stats['a_mean']

multiple_regression_model = sm.OLS(y,mutliple_x).fit()
print(multiple_regression_model.summary())
```
