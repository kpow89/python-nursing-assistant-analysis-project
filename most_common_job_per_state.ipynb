```python
df_by_state_job_stats = pd.read_csv("/Users/Kayin/Github/My_python_projects_github/Jobs Analysis 1/by_state_job_stats_bls_lowercase.csv")
```

```python
df_by_state_job_stats
```

```python
df_by_state_job_stats.info()
```

```python
#4 columns have only null vlaues so ill drop them 
df_by_state_job_stats.drop(['pct_total','pct_rpt','hourly', 'annual'], axis=1, inplace=True)
df_by_state_job_stats.info()
```

```python
#in order to get the highest number of employmen, we need ot remove the all occcupations row from each state. that row shows total employment for the whole state which is not what we want 
df_by_state_job_stats = df_by_state_job_stats[df_by_state_job_stats['jobs_1000'] != '1000.000']
```

```python
df_by_state_job_stats
```

```python
df_by_state_job_stats['tot_emp'] = pd.to_numeric(df_by_state_job_stats['tot_emp'].str.replace(',', ''), errors='coerce')
df_by_state_job_stats['tot_emp']
```

```python
df_by_state_job_stats_most_emp = df_by_state_job_stats.sort_values(by='tot_emp',ascending=False)
```

```python
df_by_state_job_stats_most_emp.head() #most common jobs in the US
```

```python
#picking the top highest job type from each state
# the problem i ran into this was every state's most employed job fell in to the same category- office! 
#so i changed it to be the top 2 highest job types
# this is done by grouping them by their state name then keeping first 2
df_top_2_highest_emp_by_state = df_by_state_job_stats_most_emp.groupby('area_title').head(2)
# df_highest_emp_by_state = df_by_state_job_stats_most_emp.drop_duplicates(subset= ['area_title'], keep='first, second')
```

```python
df_top_2_highest_emp_by_state.head()
```

```python
#b below is jsut to make sure that every state is included so i put them in order by the state name
# df_top_2_highest_emp_by_state.sort_values(by='area_title', inplace=True)
```

```python
df_top_2_highest_emp_by_state
```

```python
# then i decided to only keep the secon dhighest job, since the first is the smae for every state anywya:

df_second_highest_emp_by_state = df_top_2_highest_emp_by_state.drop_duplicates(subset= ['area_title'], keep='last')
```

```python
df_second_highest_emp_by_state
```

```python
# putting in order bystate name 
df_second_highest_emp_by_state.sort_values(by='area_title', inplace=True)
```

```python
# turnign the df to a csv so i can map it and export it 
df_second_highest_emp_by_state.to_csv("Second most common job types for each state", index=False)
```
